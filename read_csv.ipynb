This notebook has been created to allow in depth practice into the various functionalities with read_csv.

NOTE: Data used in this notebook is synthetic to allow demonstration of python functionality.

```python
import pandas as pd
from pathlib import Path
```

```python
file_path = Path(r"business_transactions_100k.csv")
output_path = Path(r"cleaned_transactions.csv")

# Remove stale output file if re-running locally 
if output_path.exists():
    output_path.unlink() #functions similar to DROP TABLE IF EXISTS in SQL

# read the CSV - note that chunksize will output a text file with a header row for each chunk, not a dataframe.
reader = pd.read_csv(file_path, chunksize=10_000)


for i, chunk in enumerate(reader):
    # Clean / transform chunk
    chunk["order_date"] = pd.to_datetime(chunk["order_date"], dayfirst=True, format="mixed", errors="coerce")
    chunk = chunk.dropna(subset=["order_date"]).copy()
    chunk["order_month"] = chunk["order_date"].dt.to_period("M").astype(str)

    # Write header only on the first chunk; append on subsequent chunks
    chunk.to_csv(
        output_path,
        mode="a",
        header=(i == 0),
        index=False,
    )
```

```python
# read the cleaned CSV to verify the output:
cleaned_df = pd.read_csv(r"cleaned_transactions.csv")

cleaned_df["order_date"] = pd.to_datetime(cleaned_df["order_date"], dayfirst=True, format="mixed",errors="coerce")

cleaned_df.info()
cleaned_df.head()
```
